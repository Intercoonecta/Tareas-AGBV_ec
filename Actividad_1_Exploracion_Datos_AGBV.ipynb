**Descripción del conjunto de datos**

El conjunto de datos utilizado corresponde a registros científicos de censos del lobo marino de Galápagos (*Zalophus wollebaeki*), realizados en la isla Caamaño, archipiélago de Galápagos, Ecuador, durante el periodo de estudio 2003–2023.

El archivo `census_avg.csv` contiene **37 registros y 25 variables**, incluyendo información sobre periodos de censo, abundancia promedio observada de machos, hembras, juveniles y crías, así como el número de conteos realizados.

La exploración inicial se efectuó mediante el lenguaje de programación R, utilizando las funciones `read.csv()`, `head()`, `dim()`, `names()` y `str()` para importar el archivo, visualizar los primeros registros e identificar la estructura de la base de datos.

Este conjunto de datos constituye una base para futuros análisis de tendencias temporales de abundancia y dinámica poblacional del lobo marino de Galápagos, así como su posible relación con la variabilidad ambiental.
